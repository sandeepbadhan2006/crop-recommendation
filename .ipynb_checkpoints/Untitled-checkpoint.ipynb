```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

import warnings
warnings.filterwarnings('ignore')
```

```python
df = pd.read_csv("crop_price_dataset.csv")
df.head()
```

```python
df.info()
```

```python
df['change'] = df['change'].fillna(df['change'].mean)
```

```python
df.info()
```

```python
df['state_name'].value_counts()
```

```python
df['month'] = pd.to_datetime(df['month'])
df['month_num'] = df['month'].dt.month
```

```python
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()
df['commodity_name'] = le.fit_transform(df['commodity_name'])
```

```python
df['change'] = df['change'].astype(float)
```

```python
df['price_range'] = df['avg_max_price'] - df['avg_min_price']
```

```python
df['price_avg'] = (df['avg_min_price'] + df['avg_max_price']) / 2
```

```python
df['last_price'] = df['avg_modal_price'].shift(1)
```

```python
df['rolling_mean'] = df['avg_modal_price'].rolling(3).mean()
```

```python
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestRegressor
```
